Japanese AI detection may ultimately depend less on grammar and more on pragmatics.

Bad English AI:
statistically predictable

Bad Japanese AI:
socially uncanny

That distinction is fascinating.

English failures often sound robotic.

Japanese failures often sound like:
“a highly intelligent exchange student who has read 40,000 pages of municipal documentation but never had a close friend.”

Which is both technically impressive and deeply eerie. Humanity continues manufacturing ghosts out of probability distributions. Tremendous species behavior.

MeCab is a Japanese morphological analyzer that breaks text into tokens. 
MeCab uses:

- dictionaries
- probabilities
- grammar models

to determine likely word boundaries and grammatical roles, since Japanese is written without spaces between words. 


This project depends on extracting linguistic features.
MeCab lets you calculate:

particle frequency
sentence-ending patterns
POS distributions
pronoun usage
connective frequency
hedging markers
vocabulary richness

Which linguistic features most clearly separate human and AI-generated Japanese?

| Feature                   | Why                      |
| ------------------------- | ------------------------ |
| Sentence length           | Easy                     |
| Connective density        | Suspected AI signal      |
| Pronoun frequency         | Suspected AI signal      |
| Kana/Kanji ratio          | Orthographic differences |
| Sentence-ending diversity | Likely strong signal     |
